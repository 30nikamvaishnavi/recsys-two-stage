# 01 - Exploratory data analysis
Run `make data` first (`MODE=sample` by default). Set `RECSYS_MODE` before launching Jupyter to change mode.

```python
import os, sys, json
sys.path.insert(0, '../src')
from pathlib import Path
from recsys.config import load_config, path, ROOT
from recsys.data.prepare import load_dataset
from recsys.eval.eda import summarize, make_figures
cfg = load_config()
ds = load_dataset(Path(path(cfg, 'data_processed')))
print(cfg['mode'], ds.stats)
```

## Summary statistics

```python
s = summarize(ds)
print(json.dumps(s, indent=2))
```

## Figures: distributions, time split, genres, cold items

```python
from IPython.display import Image, display
files = make_figures(ds, ROOT / 'docs' / 'figures' / cfg['mode'])
for f in files: display(Image(str(f)))
```

## Split sanity: stages are strictly ordered in time

```python
i = ds.interactions
i.groupby('stage')['ts'].agg(['min','max','count']).assign(min=lambda d: __import__('pandas').to_datetime(d['min'], unit='s'), max=lambda d: __import__('pandas').to_datetime(d['max'], unit='s'))
```

## Positive rate and activity by stage

```python
i.groupby('stage').agg(positive_rate=('is_pos','mean'), users=('user_idx','nunique'), items=('item_idx','nunique'))
```
